## 2.1 Dasar Teori
Python merupakan bahasa pemrograman yang memiliki berbagai macam tipe data untuk menyimpan nilai. Tipe data digunakan untuk menentukan jenis data yang akan diproses oleh program. Beberapa tipe data dasar yang digunakan dalam Python adalah integer, float, complex, boolean, dan string.

1. Integer
Integer (int) merupakan tipe data yang digunakan untuk menyimpan bilangan bulat, baik positif maupun negatif, tanpa angka desimal.
2. Float
Float (float) digunakan untuk menyimpan bilangan yang memiliki angka desimal.
3. Complex
Complex (complex) digunakan untuk menyimpan bilangan kompleks yang terdiri dari bagian real dan imajiner. Dalam Python, bilangan imajiner ditulis menggunakan huruf j.
4. Boolean
Boolean (bool) merupakan tipe data yang hanya memiliki dua nilai, yaitu True dan False. Tipe data ini biasanya digunakan untuk menyatakan suatu kondisi atau keputusan.
5. String
String (str) digunakan untuk menyimpan data berupa teks atau kumpulan karakter. String ditulis menggunakan tanda kutip tunggal (' ') atau tanda kutip ganda (" ").
6. Variabel
Variabel merupakan tempat untuk menyimpan suatu nilai dalam program. Python tidak mengharuskan programmer menentukan tipe data variabel secara langsung karena Python dapat mengenali tipe data berdasarkan nilai yang diberikan.

## 2.2 Kode Program

### 2.2.1 Program "Boolean_2511533027"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
is_lulus_3027 = True
is_cumlaude_3027 = True

#Menggunakan Boolean
nilai_3027 = 85
batas_lulus_3027 = 75

#Menentukan nilai Boolean dari kondisi
status_kelulusan_3027 = nilai_3027 >= batas_lulus_3027 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai_3027)
print("Apakah lulus?:", status_kelulusan_3027)
if is_lulus_3027 and is_cumlaude_3027:
    print("Selamat, Anda lulus dengan predikat Cumlaude!")
```

**Penjelasan :**
1. is_lulus_3027 = True dan is_cumlaude_3027 = True
→ Membuat variabel Boolean dengan nilai True yang menunjukkan kondisi lulus dan cumlaude.
2. nilai_3027 = 85 dan batas_lulus_3027 = 75
→ Menyimpan nilai mahasiswa dan batas minimal kelulusan.
3. status_kelulusan_3027 = nilai_3027 >= batas_lulus_3027
→ Membandingkan nilai dengan batas kelulusan. Jika nilai lebih besar atau sama dengan batas lulus, hasilnya True.
4. print("Apakah lulus?:", status_kelulusan_3027)
→ Menampilkan hasil pemeriksaan kelulusan.
5. if is_lulus_3027 and is_cumlaude_3027:
→ Menggunakan percabangan if dan operator and untuk memastikan kedua kondisi bernilai True.
6. print("Selamat, Anda lulus dengan predikat Cumlaude!")
→ Menampilkan pesan jika kedua kondisi terpenuhi.

### 2.2.2 Program Driver "Input_2511533027"

```python
panjang_3027 = input('Masukkan nilai panjang: ')
lebar_3027 = input('Masukkan nilai lebar: ')
luas_3027 = int(panjang_3027) * int(lebar_3027)
print("Luas =", luas_3027)
```

**Penjelasan :**
1. panjang_3027 = input(...)
→ Meminta pengguna memasukkan nilai panjang.
2. lebar_3027 = input(...)
→ Meminta pengguna memasukkan nilai lebar.
3. luas_3027 = int(panjang_3027) * int(lebar_3027)
→ Mengubah input menjadi integer, kemudian mengalikan panjang dan lebar untuk mendapatkan luas persegi panjang.
4. print("Luas =", luas_3027)
→ Menampilkan hasil perhitungan luas.

### 2.2.3 Program Driver "Konstanta_2511533027"

```python
from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_3027 = float(input('Masukkan nilai jari-jari: '))
luas_3027 = PI * jari_3027 * jari_3027
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3027, luas_3027))
```

**Penjelasan :**
1. from typing import Final
→ Mengimpor Final untuk membuat nilai yang digunakan sebagai konstanta.
2. PI: Final = 3.14
→ Menentukan nilai konstanta π sebesar 3,14.
3. jari_3027 = float(input(...))
→ Meminta pengguna memasukkan jari-jari dan mengubahnya menjadi tipe data float.
4. luas_3027 = PI * jari_3027 * jari_3027
→ Menghitung luas lingkaran menggunakan rumus π × r × r.
5. print(...)
→ Menampilkan jari-jari dan hasil luas lingkaran dengan dua angka di belakang koma.

### 2.2.4 Program Driver "numeric_2511533027"

```python
#create a variable with integer value.
a_3027=100
print("The type of variable having value", a_3027, "is", type(a_3027))

#create a variable with float value.
b_3027=10.2345
print("The type of variable having value", b_3027, "is", type(b_3027))

#create a variable with complex value.
c_3027=100+3j
print("The type of variable having value", c_3027, "is", type(c_3027))
```

**Penjelasan :**
1. a_3027 = 100
→ Membuat variabel dengan tipe data integer.
2. b_3027 = 10.2345
→ Membuat variabel dengan tipe data float.
3. c_3027 = 100 + 3j
→ Membuat variabel dengan tipe data complex, yaitu bilangan yang memiliki bagian real dan imajiner.
4. type(a_3027), type(b_3027), dan type(c_3027)
→ Digunakan untuk mengetahui tipe data dari masing-masing variabel.
