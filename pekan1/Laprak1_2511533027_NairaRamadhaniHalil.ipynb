## 2.1 Dasar Teori
Praktikum ini berfokus pada pengenalan dasar pemrograman Python, penggunaan lingkungan pengembangan Visual Studio Code, serta integrasi manajemen berkas dengan GitHub. Konsep-konsep dasar yang mendasari pelaksanaan praktikum ini meliputi:

1. Pemrograman Python dan Integrated Development Environment (IDE)
Python adalah bahasa pemrograman tingkat tinggi yang berfokus pada keterbacaan kode (readability) dan kemudahan sintaksis. Dalam praktikum ini, Visual Studio Code (VS Code) digunakan sebagai environment utama yang menyediakan fitur penyuntingan kode, eksekusi program melalui terminal terintegrasi, serta pengelolaan workspace untuk berkas berekstensi .py.
2. Pengelolaan Versi dan Repositori dengan GitHub
GitHub merupakan platform berbasis cloud yang memanfaatkan version control system (Git) untuk menyimpan, mendokumentasikan, dan memantau perubahan kode program. Integrasi direktori kerja lokal ke GitHub memudahkan pengelolaan tugas praktikum secara terstruktur dan terpusat.
3. Dokumentasi Kode Menggunakan Komentar
Komentar merupakan bagian kode yang diabaikan oleh interpreter Python saat program dijalankan dan berfungsi memberikan penjelasan atau dokumentasi. Elemen komentar dalam Python terbagi menjadi single-line comment menggunakan tanda pagar (#) dan multi-line comment yang diapit oleh tiga tanda petik ganda (""").
4. Fungsi Output dan Manipulasi String
Fungsi print() merupakan instruksi dasar untuk menampilkan keluaran teks atau data ke layar terminal. Dalam pengolahan string, terdapat elemen penting seperti operator penggabungan (+) untuk menyatukan beberapa teks, serta escape sequence seperti \n untuk membuat baris baru dan \\ untuk mencetak karakter backslash.

## 2.2 Kode Program

### 2.2.1 Program "program1.py"


```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
print("Hello World")
print()
print("Program ini menghasilkan")
print("Empat baris output")
print("/ \\ // \\\\ /// \\\\\\")
print("program ini" + " dibuat oleh a\n"
+ "naira 2511533027")
```

**Penjelasan :**
1. Baris 1 (# Program pertama)
Komentar satu baris menggunakan tanda # yang berfungsi sebagai judul program dan tidak dieksekusi oleh Python.
2. Baris 2 – 6 (""" ... """)
Komentar banyak baris yang diapit tiga tanda petik ganda untuk memberikan catatan penjelasan mengenai praktikum.
3. Baris 7 (print("Hello World"))
Mencetak teks "Hello World" ke terminal.
4. Baris 8 (print())
Membuat satu baris kosong pada tampilan output.
5. Baris 9 (print("Program ini menghasilkan"))
Mencetak teks "Program ini menghasilkan".
6. Baris 10 (print("Empat baris output"))
Mencetak teks "Empat baris output".
7. Baris 11 (print("/ \\ // \\\\\\ /// \\\\\\\\"))
Mencetak pola karakter garis miring. Penulisan \\ digunakan agar karakter backslash dapat tercetak ke layar.
8. Baris 12 – 13 (print("program ini" + ... + "naira 2511533027"))
Menggabungkan beberapa string menggunakan operator + dan menggunakan karakter \n untuk memindahkan teks nama serta NIM ke baris baru.

Tugas 1 Pekan 1


```python
# Program Tugas Pekan 1 - Profil & Kartu Nama Digital
"""Program sederhana untuk menampilkan kartu nama digital
Menggunakan print, komentar, string concatenation, dan escape character.
pertemuan 1 praktikum alpro"""
print("========================================")
print("       KARTU NAMA DIGITAL MAHASISWA     ")
print("========================================")
print()
print("Nama             : " + "Naira")
print("NIM              : " + "2511533027")
print("Program Studi    : " + "Informatika")
print("Fakultas         : " + "Fakultas Teknologi Informasi")
print("Kelas            : " + "Prak Alpro A")
print()
print("  /\\_/\\  ")
print(" ( o.o ) ")
print("  > ^ <  ")
print("========================================")
print("Program ini dibuat oleh:\nNaira - 2511533027")
```

Program Python di atas dibuat untuk menampilkan kartu nama digital mahasiswa menggunakan konsep-konsep dasar pemograman Python seperti fungsi print(), komentar, string concatenation, dan escape character. Pada bagian awal, komentar satu baris (#) dan docstring ("""...""") digunakan untuk memberikan informasi mengenai judul dan deskripsi program tanpa memengaruhi jalannya kode. Selanjutnya, fungsi print() digunakan secara berurutan untuk mencetak garis dekoratif, baris kosong sebagai jarak, serta identitas mahasiswa yang digabungkan menggunakan operator + (string concatenation).

Selain menampilkan teks biasa, program ini juga menampilkan ASCII art berbentuk wajah kucing serta keterangan pembuat program di bagian akhir. Pada cetakan ASCII art, digunakan escape character \\ (double backslash) agar simbol backslash dapat ditampilkan sebagai teks biasa. Terakhir, karakter \n (newline) digunakan di dalam satu fungsi print() untuk memindahkan baris secara otomatis, sehingga tampilan pembuat program menjadi lebih rapi dan terstruktur di layar output.
